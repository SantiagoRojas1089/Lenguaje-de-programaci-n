## Paso 6 — opción 1: registrar una factura con `for`

Completa la opción `1` del menú. Necesitas, en este orden:

1. Preguntar cuántos productos tiene la factura (`cantidad`, convertida a entero con `int()`)
2. Crear `total_factura` empezando en 0
3. Un `for i in range(cantidad):` que en cada vuelta pida nombre y precio, y sume el precio a `total_factura`
4. Imprimir el total de la factura
5. Sumar `total_factura` a `total_dia`
6. Sumar 1 a `facturas_hechas`

Recuerda poner `total_dia = 0` y `facturas_hechas = 0` **antes** del `while` (paso 5), no adentro de la opción 1.

````python
total_dia = 0
facturas_hechas = 0

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        # ✏️ 1. Pregunta cuántos productos tiene la factura
        cantidad = int(input("¿Cuántos productos tiene esta factura? "))

        # ✏️ 2. Crea total_factura empezando en 0
        total_factura = 0

        # ✏️ 3. for i in range(cantidad): pide nombre y precio, suma precio a total_factura
        for i in range(cantidad):
            print(f"Producto {i + 1} de {cantidad}:")
            nombre = input("  Nombre del producto: ")
            precio = float(input("  Precio: "))
            total_factura = total_factura + precio

        # ✏️ 4. Imprime el total de la factura
        print(f"El total de esta factura es: ${total_factura}")

        # ✏️ 5. Suma total_factura a total_dia
        total_dia = total_dia + total_factura

        # ✏️ 6. Suma 1 a facturas_hechas
        facturas_hechas = facturas_hechas + 1

    elif opcion == "2":
        print(f"El total acumulado del día es: ${total_dia}")

    elif opcion == "3":
        print(f"Se han registrado {facturas_hechas} facturas hoy.")

    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break

    else:
        print("Opción no válida, intenta de nuevo.")


````

## Misión 3 — facturación completa (un `for` sobre las líneas, encabezado en tupla)

Ahora sí completamos de verdad la opción `1` del menú. En vez de pedir nombre y precio sueltos, la factura va a usar `catalogo` y `clientes`. Necesitas, en este orden:

1. Pedir el **documento** del cliente y buscarlo en `clientes`. Si no existe, avisa y sigue con un cliente genérico (`"Consumidor final"`).
2. Pedir **cuántas líneas** tiene la factura (`cantidad_lineas`, con `int()`).
3. Sumar 1 a `numero_factura` y armar el **encabezado como tupla**: `encabezado = (numero_factura, documento, nombre_cliente)` — son datos que no deberían cambiar una vez emitida la factura.
4. Un `for i in range(cantidad_lineas):` que en cada vuelta:
   - pida el **código** del producto
   - lo busque en `catalogo` (si no existe, avisa y salta a la siguiente línea con `continue`)
   - pida la **cantidad**
   - calcule el subtotal de esa línea **con IVA incluido**: `precio * cantidad * (1 + iva)`
   - lo sume a `total_factura`
5. Imprimir el encabezado y el total de la factura.
6. Sumar `total_factura` a `total_dia` y 1 a `facturas_hechas`, igual que antes.

````python
total_dia = 0
facturas_hechas = 0
numero_factura = 1000

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        documento = input("Documento del cliente: ").strip()

        if documento in clientes:
            nombre_cliente = clientes[documento]["nombre"]
        else:
            print("  ⚠️ Cliente no encontrado, se registra como consumidor final.")
            nombre_cliente = "Consumidor final"

        cantidad_lineas = int(input("¿Cuántas líneas tiene esta factura? "))

        numero_factura = numero_factura + 1
        encabezado = (numero_factura, documento, nombre_cliente)

        total_factura = 0

        for i in range(cantidad_lineas):
            codigo = input(f"  Código del producto (línea {i + 1} de {cantidad_lineas}): ").strip().upper()

            if codigo not in catalogo:
                print("  ⚠️ Producto no encontrado en el catálogo, se omite esta línea.")
                continue

            producto = catalogo[codigo]
            cantidad = int(input(f"  Cantidad de '{producto['nombre']}': "))
            subtotal_linea = producto["precio"] * cantidad * (1 + producto["iva"])
            total_factura = total_factura + subtotal_linea
            print(f"  ✅ {cantidad} x {producto['nombre']} = ${subtotal_linea:,.0f}")

        print(f"Factura N° {encabezado[0]} — Cliente: {encabezado[2]} (doc. {encabezado[1]})")
        print(f"Total de la factura: ${total_factura:,.0f}")

        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1

    elif opcion == "2":
        print(f"El total acumulado del día es: ${total_dia:,.0f}")
    elif opcion == "3":
        print(f"Se han registrado {facturas_hechas} facturas hoy.")
    elif opcion == "4":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
````

````python
catalogo = {
    "P-001": {"nombre": "Cuaderno cuadriculado", "precio": 8500,  "iva": 0.19},
    "E-011": {"nombre": "Caja de lapiceros x12", "precio": 15990, "iva": 0.19},
    "L-030": {"nombre": "Libro juvenil",         "precio": 45000, "iva": 0.00},
    "R-050": {"nombre": "Revista educativa",     "precio": 12000, "iva": 0.05},
}

clientes = {
    "1001234567":  {"nombre": "Juan Pérez",        "tipo": "particular"},
    "1029876543":  {"nombre": "Camila Restrepo",   "tipo": "estudiante"},
    "900123456-1": {"nombre": "Colegio San José",  "tipo": "colegio"},
}

inventario = {
    "P-001": 30,
    "E-011": 15,
    "L-030": 10,
    "R-050": 20,
}

total_dia = 0
facturas_hechas = 0
numero_factura = 1000

while True:
    print("-" * 40)
    print("EL LÁPIZ FELIZ")
    print("-" * 40)
    print("1. Registrar una factura")
    print("2. Ver total acumulado del día")
    print("3. Ver cuántas facturas se han registrado")
    print("4. Consultar inventario")
    print("5. Salir")
    print("-" * 40)

    opcion = input("Elige una opción: ")

    if opcion == "1":
        documento = input("Documento del cliente: ").strip()

        if documento in clientes:
            nombre_cliente = clientes[documento]["nombre"]
        else:
            print("  ⚠️ Cliente no encontrado, se registra como consumidor final.")
            nombre_cliente = "Consumidor final"

        cantidad_lineas = int(input("¿Cuántas líneas tiene esta factura? "))

        numero_factura = numero_factura + 1
        encabezado = (numero_factura, documento, nombre_cliente)

        total_factura = 0

        for i in range(cantidad_lineas):
            codigo = input(f"  Código del producto (línea {i + 1} de {cantidad_lineas}): ").strip().upper()

            if codigo not in catalogo:
                print("  ⚠️ Producto no encontrado en el catálogo, se omite esta línea.")
                continue

            producto = catalogo[codigo]
            cantidad = int(input(f"  Cantidad de '{producto['nombre']}': "))
            subtotal_linea = producto["precio"] * cantidad * (1 + producto["iva"])
            total_factura = total_factura + subtotal_linea
            print(f"  ✅ {cantidad} x {producto['nombre']} = ${subtotal_linea:,.0f}")

            if codigo not in inventario:
                print("  ⚠️ Este producto no tiene registro de inventario.")
            else:
                inventario[codigo] = inventario[codigo] - cantidad

        print(f"Factura N° {encabezado[0]} — Cliente: {encabezado[2]} (doc. {encabezado[1]})")
        print(f"Total de la factura: ${total_factura:,.0f}")

        total_dia = total_dia + total_factura
        facturas_hechas = facturas_hechas + 1

    elif opcion == "2":
        print(f"El total acumulado del día es: ${total_dia:,.0f}")
    elif opcion == "3":
        print(f"Se han registrado {facturas_hechas} facturas hoy.")
    elif opcion == "4":
        codigo_consulta = input("Código a consultar: ").strip().upper()
        if codigo_consulta in inventario:
            print(f"Quedan {inventario[codigo_consulta]} unidades de '{catalogo[codigo_consulta]['nombre']}' en bodega.")
        else:
            print("  ⚠️ Ese código no existe en el inventario.")
    elif opcion == "5":
        print("¡Gracias por usar el sistema! Hasta pronto.")
        break
    else:
        print("Opción no válida, intenta de nuevo.")
````
